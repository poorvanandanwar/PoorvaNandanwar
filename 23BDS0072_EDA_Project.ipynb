```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1.Loading the dataset
df = pd.read_csv("wages.csv")
df.head()
```

```python
# 3.Handling missing data
df.isnull().sum()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
df.isnull().sum()
```

```python
# 4.Data cleaning
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
#5.Data transformation
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in df.select_dtypes(include='object'):
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numeric = df.select_dtypes(include=np.number)
df[numeric.columns] = scaler.fit_transform(numeric)
```

```python
#6.Univariate analysis with a minimum of three visualizations
#Histogram
df.hist(figsize=(12,8))
plt.show()
```

```python
#boxplot
plt.figure(figsize=(12,8))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
#violinplot
plt.figure(figsize=(12,8))
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
#7.Bivariate analysis with a minimum of three visualizations
#scatter plot
sns.scatterplot(
    x='ue.mean',
    y='ue1',
    data=df
)
plt.show()
```

```python
#correlation heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(),
            annot=True,
            cmap='coolwarm')
plt.show()
```

```python
#boxplot
plt.figure(figsize=(10,6))
sns.boxplot(data=df[['lnw', 'exper', 'postexp', 'uerate']])
plt.title("Comparison of Important Numerical Variables")
plt.xticks(rotation=45)
plt.show()
```

```python
#8.Multivariate analysis with a minimum of three visualizations
#bubbleplot
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='exper',
    y='lnw',
    size='postexp',
    hue='uerate',
    sizes=(20,300),
    alpha=0.7
)
plt.title("Experience vs Log Wage")
plt.show()
```

```python
#correlation heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(
    df.corr(),
    annot=True,
    cmap="viridis"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
#scatterplot with hue
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df,
    x='exper',
    y='lnw',
    hue='black',
    palette='Set1'
)
plt.title("Experience vs Log Wage by Race")
plt.show()
```
